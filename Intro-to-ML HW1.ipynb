```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import drive
drive.mount('/content/drive')
```

```python
file_path = '/content/drive/My Drive/Courses/Intro-to-ML/Datasets/D3.csv'
D3 = pd.DataFrame(pd.read_csv(file_path))
D3.head()
```

```python
D3.isnull().sum()*100/D3.shape[0]
```

```python
X1 = D3[['X1']].to_numpy(dtype=float)
X2 = D3[['X2']].to_numpy(dtype=float)
X3 = D3[['X3']].to_numpy(dtype=float)

Y = D3[['Y']].to_numpy(dtype=float)

learning_rate = 0.01
max_iterations = 1000
m = len(X1)

# FOR DATA SET X1
theta1 = np.zeros((2,1))
X1_b = np.c_[np.ones((len(X1), 1)), X1]
train_loss1 = []


for iteration in range(max_iterations):
    gradients1 = 2/m * X1_b.T.dot(X1_b.dot(theta1) - Y)
    theta1 -= learning_rate * gradients1

    loss1 = (1/m) * np.sum(np.square(X1_b.dot(theta1) - Y))

    train_loss1.append(loss1)

# FOR DATA SET X2
theta2 = np.zeros((2,1))
X2_b = np.c_[np.ones((len(X2), 1)), X2]
train_loss2 = []


for iteration in range(max_iterations):
    gradients2 = 2/m * X2_b.T.dot(X2_b.dot(theta2) - Y)
    theta2 -= learning_rate * gradients2

    loss2 = (1/m) * np.sum(np.square(X2_b.dot(theta2) - Y))

    train_loss2.append(loss2)

# FOR DATA SET X3
theta3 = np.zeros((2,1))
X3_b = np.c_[np.ones((len(X3), 1)), X3]
train_loss3 = []


for iteration in range(max_iterations):
    gradients3 = 2/m * X3_b.T.dot(X3_b.dot(theta3) - Y)
    theta3 -= learning_rate * gradients3

    loss3 = (1/m) * np.sum(np.square(X3_b.dot(theta3) - Y))

    train_loss3.append(loss3)
```

```python
#TRAINING LOSS, INDIVIDUAL FEATURES
plt.plot(train_loss1, label="Training Loss, X1")
plt.xlabel("Iteration")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.title("Training Loss for X1 Over Iterations")
plt.show()

plt.plot(train_loss2, label="Training Loss, X2")
plt.xlabel("Iteration")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.title("Training Loss for X2 Over Iterations")
plt.show()

plt.plot(train_loss3, label="Training Loss, X3")
plt.xlabel("Iteration")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.title("Training Loss for X3 Over Iterations")
plt.show()
```

```python
prediction1 = X1_b.dot(theta1)
prediction2 = X2_b.dot(theta2)
prediction3 = X3_b.dot(theta3)

plt.plot(X1[:, 0], prediction1[:, 0], color="red", label="Regression Model")
plt.scatter(X1[:, 0], Y[:, 0], label="Original Data")
plt.xlabel("X1")
plt.ylabel("Y")
plt.legend()
plt.title("Final Regression Model Using X1")
plt.show()

plt.plot(X2[:, 0], prediction2[:, 0], color="red", label="Regression Model")
plt.scatter(X2[:, 0], Y[:, 0], label="Original Data")
plt.xlabel("X2")
plt.ylabel("Y")
plt.legend()
plt.title("Final Regression Model Using X2")
plt.show()

plt.plot(X3[:, 0], prediction3[:, 0], color="red", label="Regression Model")
plt.scatter(X3[:, 0], Y[:, 0], label="Original Data")
plt.xlabel("X3")
plt.ylabel("Y")
plt.legend()
plt.title("Final Regression Model Using X3")
plt.show()
```

```python
X_all = D3[['X1', 'X2', 'X3']].to_numpy(dtype=float)
Y = D3[['Y']].to_numpy(dtype=float)

learning_rate = 0.01
max_iterations = 1000
m = len(Y)

# DATASETS X1, X2, X3
X_all_b = np.c_[np.ones((m, 1)), X_all]
theta_all = np.zeros((4, 1))
train_loss_all = []

for iteration in range(max_iterations):

    predictions = X_all_b.dot(theta_all)
    errors = predictions - Y

    gradients_all = (2 / m) * X_all_b.T.dot(errors)

    theta_all -= learning_rate * gradients_all

    loss_all = (1 / m) * np.sum(
        np.square(X_all_b.dot(theta_all) - Y)
    )

    train_loss_all.append(loss_all)
```

```python
# TRAINING LOSS FOR ALL FEATURES
plt.plot(train_loss_all, label="Training Loss X1, X2, X3")
plt.xlabel("Iteration")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.title("Training Loss for X1, X2, and X3 Over Iterations")
plt.show()
```
